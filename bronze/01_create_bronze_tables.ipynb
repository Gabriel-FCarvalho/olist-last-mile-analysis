# Bronze Layer — Delta Tables Setup

Reads raw CSV files from the Volume `/Volumes/workspace/olist_raw/bronze/` and registers each one as a Delta table under the `workspace.bronze` schema.

**Design decision:** setting `multiLine = true` was required for `order_reviews` due to line breaks in the comment fields of the original CSV. Without this parameter, the parser misaligned columns and inflated the row count from 99,249 to 104,162.

These Delta tables are the read source for the Silver layer.

```sql
%sql
CREATE SCHEMA IF NOT EXISTS workspace.bronze;

CREATE TABLE IF NOT EXISTS workspace.bronze.orders
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_orders_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.customers
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_customers_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.order_items
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_order_items_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.order_payments
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_order_payments_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.order_reviews
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_order_reviews_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true',
  multiLine => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.products
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_products_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.sellers
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_sellers_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.geolocation
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/olist_geolocation_dataset.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);

CREATE TABLE IF NOT EXISTS workspace.bronze.product_category_translation
USING DELTA
AS SELECT * FROM read_files(
  '/Volumes/workspace/olist_raw/bronze/product_category_name_translation.csv',
  format => 'csv', header => 'true', inferSchema => 'true'
);
```
